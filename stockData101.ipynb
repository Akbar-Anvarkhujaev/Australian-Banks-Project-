```python
%pip install yfinance
%pip install matplotlib
import datetime as dt 
import pandas as pd 
import yfinance as yf 
```

```python
end = dt.datetime.now()
start = end - dt.timedelta(days=5000)
start, end
```

```python
stocklist= ['CBA','NAB','WBC','ANZ']
stocks = [i+ '.AX' for i in stocklist]
stocks
```

```python
df = yf.download(stocks, start=start, end=end, group_by="ticker", auto_adjust=False)
df.head() 
```

```python
df.index
```

```python
df.columns
```

```python
close = df.xs('Close', axis=1, level=1)
close.head()
```

```python
close[close.index > end - dt.timedelta(days=100)].describe()
```

```python
close.plot(figsize=(12,8))
```
